```python
import random

arr = [random.randint(0, 50) for _ in range(20)]
print("Вихідний масив:", arr)

threshold = int(input("Введіть порогове значення: "))

count_above = sum(1 for x in arr if x > threshold)
print(f"Кількість елементів > {threshold}: {count_above}")

max_val = max(arr)
max_idx = arr.index(max_val)
print(f"Максимальне значення: {max_val}, позиція першої появи: {max_idx}")

sorted_arr = sorted(arr, reverse=True)
print("Відсортований масив (за спаданням):", sorted_arr)
```

```python
import random

low = int(input("Введіть нижню межу: "))
high = int(input("Введіть верхню межу: "))

matrix = [[random.randint(low, high) for _ in range(5)] for _ in range(5)]

print("\nВихідна матриця 5x5:")
for row in matrix:
    print(row)

diag = [matrix[i][i] for i in range(5)]
print("\nЕлементи головної діагоналі:", diag)
print("Сума елементів головної діагоналі:", sum(diag))

modified_matrix = []
for i in range(5):
    new_row = []
    for j in range(5):
        if j > i:  
            new_row.append(0)
        else:
            new_row.append(matrix[i][j])
    modified_matrix.append(new_row)

print("\nМатриця після обнулення вище діагоналі:")
for row in modified_matrix:
    print(row)
```

```python

start = int(input("Введіть початок діапазону: "))
end = start + 30

seq = list(range(start, end))
matrix = [seq[i * 5:(i + 1) * 5] for i in range(6)]

print("\nМатриця 6x5:")
for row in matrix:
    print(row)

row_sums = [sum(row) for row in matrix]
print("\nСума елементів у кожному рядку:", row_sums)

col_maxes = [max(matrix[row][col] for row in range(6)) for col in range(5)]
print("Максимуми за стовпчиками:", col_maxes)
```

```python
import random

low = int(input("Введіть нижню межу (включно з від'ємними): "))
high = int(input("Введіть верхню межу: "))

arr = [random.randint(low, high) for _ in range(15)]
print("\nВихідний масив:", arr)

negatives = [x for x in arr if x < 0]
print("Від'ємні елементи:", negatives)

arr_copy = [x if x >= 0 else 0 for x in arr]
print("Змінений масив (від'ємні замінено на 0):", arr_copy)

zero_count = arr_copy.count(0)
print("Кількість нульових елементів у зміненому масиві:", zero_count)
```

```python
import random

length = int(input("Введіть довжину масивів: "))

arr1 = [random.randint(0, 10) for _ in range(length)]
arr2 = [random.randint(10, 20) for _ in range(length)]

print("\nПерший масив (0-10):", arr1)
print("Другий масив (10-20):", arr2)

concatenated = arr1 + arr2
print("Об'єднаний масив:", concatenated)

added = [arr1[i] + arr2[i] for i in range(length)]
subtracted = [arr1[i] - arr2[i] for i in range(length)]

print("Поелементне додавання:", added)
print("Поелементна різниця:", subtracted)
```
