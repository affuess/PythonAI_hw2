```python
import pandas as pd

```

```python
df = pd.read_csv('coffee_orders.csv')
print("First 10 strings")
display(df.head(10))
print(" \nInto about dataset")
df.info()
```

```python
df_clean = df.dropna(subset = ['Напій', 'Ціна']).copy()

df_clean['Кількість'] = df_clean['Кількість'].fillna(1)
df_clean = df_clean.drop_duplicates()
df_clean['Кількість'] = df_clean['Кількість'].astype(int)
print(f"Count lines: {len(df_clean)}")
display(df_clean.head())
```

```python
df_clean['Виручка'] = df_clean['Ціна'] * df_clean['Кількість']
print("Statistic")
display(df_clean.describe())
```

```python
heigh_revenue_orders = df_clean[df_clean['Виручка']>150]
print(f'Count big orders: {len(heigh_revenue_orders)}')
display(heigh_revenue_orders.head())
kyiv_orders = df_clean[df_clean['Місто'] == 'Київ'].sort_values(by="Виручка", ascending=False)
display(kyiv_orders.head())
```

```python
city_report = df_clean.groupby('Місто').agg(
    Загальна_Виручка = ('Виручка', 'sum'),
    Всього_порцій = ('Кількість', 'mean'),
    Середній_чек = ('Виручка', 'mean')
).reset_index()
city_report = city_report.sort_values(by='Загальна_Виручка', ascending=False)
print("Finance report")
display(city_report)
```
